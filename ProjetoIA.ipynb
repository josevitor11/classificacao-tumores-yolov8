```python
#Imports principais
from ultralytics import YOLO
from pathlib import Path
from pathlib import Path
from ultralytics import YOLO
import yaml
```

```python
#====== Ajuste o caminho do seu dataset no Frive ======
DATASETS_DIR = Path("/content/drive/MyDrive/mini_split") # <- ajuste se o nome

#Lista de classes (altere se necessário)
names = ["Glioma","Meningioma","No Tumor","Pituitary"]

#Criação do dicionário YAML
data_yaml = {
    "path": str(DATASETS_DIR),
    "train": "train", #pasta relativas a DATASET_DIR
    "val": "val",
    "test":"test",
    "names": names

}

#Salva o arquivo dentro do dataset
yaml_path = DATASETS_DIR / "data.yaml"
with open(yaml_path, "w") as f:
  yaml.safe_dump(data_yaml, f, sort_keys=False, allow_unicode=True)

print("data.yaml criado em:",yaml_path)
print(yaml.safe_dump(data_yaml,sort_keys=False, allow_unicode=True))
```

```python
DATASETS_DIR = Path("/content/drive/MyDrive/mini_split")

train_path = DATASETS_DIR / "train"
val_path = DATASETS_DIR / "val"
test_path = DATASETS_DIR / "test"
yaml_path = DATASETS_DIR / "data.yaml"

print("Dataset path definidos:")
print("Train:",train_path)
print("Val:", val_path)
print("Test:", test_path)
print("YAML:", yaml_path)
```

```python
model = YOLO("yolov8n.pt")
```

```python
results = model.train(
    data=str(yaml_path),
    epochs=100,
    imgsz=640,
    batch=16,
    patience=15,
    pretrained=True,
    workers=2,
    val=True
)

results
```
